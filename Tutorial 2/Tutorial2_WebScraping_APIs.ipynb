### Web Scraping: extrayendo datos de internet

#### ¿Qué es web scraping?

La práctica de recopilar datos a través de cualquier medio que no sea un programa que interactúa con una API o un humano que usa un navegador web. En general esto se logra mediante un programa automatizado que consulta un servidor web, solicita datos (generalmente en forma de HTML y otros archivos que componen las páginas web) y luego analiza esos datos para extraer la información necesaria.

```python
url = "https://www.lanacion.com.ar/"

r = requests.get(url) #traigo el contenido del html
contenido = r.content

soup = BeautifulSoup(contenido, "html.parser")
soup
```

### Nocturna, diminuta y “cafetera”: descubren una nueva especie de rana en Costa Rica

### Como encontrar secciones o texto especifico

Usando find y find_all

```python
# Dentro de la sopa, busco los elementos que contienen la información que necesito
# Buscamos el elemento <h2> indicando la clase (class). Escribo class_ porque "class" es una palabra reservada en Python
h2_element = soup.find('h2', class_='text ln-text title --prumo --font-medium --font-m-l')
print(h2_element)
```

```python
# Obtenemos el texto del elemento <h2>
h2_text = h2_element.text.strip()  # strip() permite remover espacios sobrantes
print('\n', h2_text)
```

```python
# Obtuvimos el *primer* elemento de la página con ese tag. Pero queremos hacerlo para todos los títulos...
# El método "find_all" busca TODOS los elementos de la página con ese tag y devuelve una lista que los contiene
# (en realidad devuelve un objeto de la clase "bs4.element.ResultSet")
h2_elements = soup.find_all('h2')

print(type(h2_elements))
print('\n', h2_elements)
```

```python
# Extraemos el texto de cada elemento <h2> e imprimimos
for h2_element in h2_elements:
    h2_text = h2_element.text.strip()
    print(h2_text)
```

```python
# Idealmente, tenemos que guardar estos títulos, queremos analizarlos
titulos = [] # primero creamos una lista

# Extraemos el texto de cada elemento <h2> y ahora guardamos
for h2_element in h2_elements:
    h2_text = h2_element.text.strip()
    #print(h2_text)

    titulos.append({
        'titular': h2_text
    })

# Creamos un dataframe a partir de la lista de títulos
titulos_df = pd.DataFrame(titulos)
```

### Otro Ejemplo de Web Scraping: Tablas y como *disfrazar* nuestra actividad

Vamos a _scrappear_ una tabla de episodios de Game of Thrones de la lista de episodios que aparece en Wikipedia (https://en.wikipedia.org/wiki/List_of_Game_of_Thrones_episodes). Les recomiendo que entren al link para ver más o menos la estructura de lo que queremos _scrappear_. Lo que vamos a tratar de obtener es una lista con todos los episodios, su fecha de emisión, director, escritor y cantidad de personas que lo vierons.

Para eso vamos a utilizar una función de pandas que nos permite leer html: `read_html`

Estén seguros de tener instalado el módulo `lxml`. Lo pueden instalar con `conda install lxml`, con `pip install lxml` o con el instalador de módulos gráfico del Anaconda Navigator.


```python
url = 'https://en.wikipedia.org/wiki/List_of_Game_of_Thrones_episodes'
```

Ahora que tenemos la dirección en la variable `html` vamos a usar la función `read_html`:

```python
table = pd.read_html(url)
print(len(table))
```

**¿Qué pasó?**

Básicamente, Wikipedia detecto que un *script automático* de Python estaba intentando acceder a su pagina web. Para protegerse de ataques de DDoS o que bots sobrecarguen sus servidores, optan por bloquear cualquier petición que no provenga de un navegador real.

Para solucionar esto, tenemos que disfrazar nuestra actividad usando un encabezado llamado ***User-Agent***:

```python
from io import StringIO

#Creamos un "disfraz" (User-Agent) simulando ser Google Chrome en Windows
encabezados = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"
}

#Hacemos la petición a la página usando nuestro encabezado
respuesta = requests.get(url, headers=encabezados)

#Le pasamos el texto HTML descargado a Pandas usando StringIO
tablas = pd.read_html(StringIO(respuesta.text))

#Vemos cuantas tablas tiene
print(f"Se encontraron {len(tablas)} tablas.")
```

Vemos que table tiene 22 elementos. Veamos el elemento 0.

```python
print(tablas[0])
```

Se ve que es la primera tabla que aparece en la página web, una tabla que contiene la cantidad de episodios por temporada. Entonces busquemos la primera y la última de las tablas que nos interesan.

```python
print(tablas[1])
```

```python
print(tablas[8])
```

Después ya podemos ver que las tablas siguientes con de cosas que no nos interesan.

```python
print(tablas[9])
```

Armemos un subset con las tablas que nos interesan:

```python
subset = tablas[1:9]
print(subset)
```

Ahora podemos directamente concatenar todas las tablas (una de las ventajas de usar el _parser_ de `pandas` es que las tablas ya son dataframes).

```python
df_total = pd.concat(subset)
display(df_total)
```

Pero qué pasa si queremos agregar una columna que contenga el número de temporada.

```python
for i in range(len(subset)):
    subset[i]['season'] = i + 1
```

```python
df_total = pd.concat(subset)
display(df_total)
```

Y por último, movamos la columna season al principio:

```python
columns = ['season'] + [col for col in df_total.columns if col != 'season']
print(columns)
```

```python
df_total = df_total[columns]
df_total.head()
```

Listo, ahora los podemos exportar como archivo `csv`:

```python
df_total.to_csv('lista_GoT.csv', index=False)
```

### APIs: *pidiendo* datos de internet

¿Qué es una API? Los sistemas tienen distintos tipos de interfaces que permiten interactuar con ellos.

* **GUI (Graphical User Interface o interfaz gráfica)**: El usuario clickea e interactúa con distintos objetos para ejecutar acciones y lograr sus objetivos. Las páginas web tienen interfaces gráficas.
* **API (Application Programming Interface o interfaz de programación)**: El usuario escribe líneas de código para interactuar con el sistema, ejecutar acciones y lograr sus objetivos.

Todos los sitios web tienen una interfaz gráfica con la que estamos acostumbrados a interactuar, y también tienen una interfaz de programación más o menos desarrollada, o más o menos expuesta, con la cual tal vez no estemos tan acostumbrados a interactuar.

Ejemplo:

En https://www.mercadolibre.com.ar/ podemos poner en el buscador "pelotas", apretar "Enter" y el sistema devuelve el resultado de una búsqueda. Si, en cambio, ponemos "pelotas futbol" y apretamos "Enter" de nuevo, cambiará el resultado de la búsqueda. Esta es la **interfaz gráfica del sitio**.

Sin embargo, también podríamos lograr el mismo objetivo sólo escribiendo distintas URLs en el navegador:

* https://listado.mercadolibre.com.ar/pelotas
* https://listado.mercadolibre.com.ar/pelotas-futbol

Esto es más parecido a lo que llamaríamos una **interfaz de programación**.


Las APIs son mecanismos que permiten que una aplicación o servicio acceda a un recurso en otra aplicación o servicio (un cliente accede a recurso en un servidor).

En la web, las interfaces de programación de uso más difundido son las APIs REST ([Representational state transfer](https://es.wikipedia.org/wiki/Transferencia_de_Estado_Representacional)). REST es un estilo para diseñar aplicaciones que especifica un conjunto de protocolos y métodos para interactuar con los recursos de internet escribiendo líneas de código.

Estas APIs se comunican mediante el protocolo HTTP para interactuar con los recursos (es el medio de comunicación comunicación entre el usuario y el servidor). HTTP es el mismo protocolo que se utiliza para cargar páginas web en un navegador.
Este protocolo indica cómo estructurar un mensaje de texto que describa la petición (**request**) del usuario o cliente a un servidor. Hay distintos tipos de peticiones que un usuario puede realizar, algunas de ellas son:

* **POST**: Envía datos al servidor para crear un recurso nuevo.
* **GET**: Se utiliza para obtener información de un recurso alojado en el servidor.
* **PUT**: Actualiza (crea o modifica) un recurso del servidor.
* **DELETE**: Elimina un recurso del servidor.

El cliente manda una petición (**request**) y el servidor devuelve una respuesta (**response**).

Cada vez que vamos al navegador y escribimos la dirección de una página web, **estamos haciendo un GET request** a un servidor. Esto es una petición para adquirir el código de un recurso que queremos visualizar en el navegador.

La URL es la parte más importante de la definición de un GET request (aunque el navegador agrega otras cosas también, que no vemos) y nos permite cambiar la representación deseada de un mismo recurso de distintas maneras:

* https://deportes.mercadolibre.com.ar/pelotas-futbol pide al servidor pelotas de fútbol.
* https://deportes.mercadolibre.com.ar/pelotas-futbol_OrderId_PRICE pide al servidor pelotas de fútbol ordenadas por precio.

Cuando escribimos una URL en un navegador, la mayoría de las veces hacemos GET requests que devuelven código HTML (el código que da una estructura a una página web, tal como cuando obteníamos el código HTML al hacer web scraping). Pero los GET requests pueden devolver datos en otros formatos (por ejemplo en JSON y en CSV).

Las APIs REST que definen GET requests capaces de devolver datos en formato JSON y CSV, son particularmente útiles cuando queremos analizar datos.

#### Obtener ítems de una consulta de búsqueda

- https://developers.mercadolibre.com.ar/es_ar/items-y-busquedas

Con el parámetro “q” puedes realizar una búsqueda por palabra, frases o tributos claves de búsqueda. Es importante tener en cuenta que en este campo puedes enviar tantos detalles de los atributos de la publicación como sea necesario, de esta manera en la respuesta obtendrás búsquedas más precisas. Si quieres separar términos puedes usar %20 que corresponde al código ASCII.


```python
#Así se ve la búsqueda en el navegador:
# https://pokeapi.co/api/v2/pokemon

#Al usar la API podemos agregar distintos parametros como limit (cantidad) y offset (desde donde):
# https://pokeapi.co/api/v2/pokemon?limit=10&offset=0

url = "https://pokeapi.co/api/v2/pokemon?limit=5"
print(url)

# Hacemos el pedido o request y obtenemos la response
response = requests.request("GET", url)
print(response) # correcto

# Vemos el texto
response.text
```

```python
#Podemos guardar el resultado en un diccionario
data = response.json()
data
```

```python
#Ahora podemos acceder a los distintos resultados (usando las keys del diccionario)

#Veamos la longitud de la lista de resultados
print('Cantidad de Pokémon:', len(data['results']))

#Veamos el nombre del primer Pokémon
print('Nombre del primer Pokémon:', data['results'][0]['name'])

#Link para obtener los detalles del primer Pokémon
print('URL de detalles:', data['results'][0]['url'])
```

#### Consultar detalles de un Pokémon específico

Usando la URl que obtuvimos en el paso anterior, podemos consultar datos específicos como tipo, altura y peso.  

```python
#Al usar la API tenemos que seguir este formato:
#https://pokeapi.co/api/v2/pokemon/{nombre_o_id}

#Veamos un ejemplo buscando los detalles de Bulbasaur
nombre_pokemon = data['results'][0]['name'] #Esto es "bulbasaur"

url_detalle = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}" #Reemplazamos por el nombre
response_detalle = requests.get(url_detalle)

#Guardamos el JSON
datos_pokemon = response_detalle.json()

#Imprimimos información específica
print(f"Nombre: {datos_pokemon['name'].capitalize()}")
print(f"Altura: {datos_pokemon['height'] / 10} metros")
print(f"Peso: {datos_pokemon['weight'] / 10} gramos")

#La información de los tipos viene en una lista anidada
tipo = datos_pokemon['types'][0]['type']['name']
print(f"Tipo principal: {tipo}")
```

### API Series de Tiempo

La **[API Series de Tiempo de la Republica Argentina](https://apis.datos.gob.ar/series)** es una API REST desarrollada y mantenida por el Estado Nacional de Argentina para la consulta de estadísticas en formato de series de tiempo. Contiene series publicadas por organismos de la Administración Pública Nacional.

La API permite:

* [Buscar series](https://datosgobar.github.io/series-tiempo-ar-api/reference/search-reference/) por texto. También se pueden buscar en el sitio web: https://datos.gob.ar/series
* Cambiar la frecuencia (por ejemplo: convertir series diarias en mensuales)
* Elegir la función de agregacion de valores, usada en el cambio de frecuencia (una serie se puede convertir de diaria a mensual promediando, sumando, sacando el maximo, el minimo, el ultimo valor del periodo, etc)
* Filtrar por rango de fechas
* Elegir el formato (CSV o JSON)
* Cambiar configuracion del CSV (caracter separador, caracter decimal)

En https://datos.gob.ar/series se puede buscar series de tiempo publicadas por distintos organismos de la Administración Pública Nacional en Argentina y usar el link al CSV para leerlos directamente desde python con pandas.

También podes **buscar los ids de las series de interés** y juntarlos en la misma consulta para armar una tabla de hasta 40 series.

```python
# Un ejemplo

url_arg = "https://apis.datos.gob.ar/series/api/series?ids=102.1_I2NG_ABRI_M_22,101.1_I2NG_2016_M_22&format=json&start_date=2016-01-01&collapse=year&collapse_aggregation=avg](https://apis.datos.gob.ar/series/api/series?ids=102.1_I2NG_ABRI_M_22,101.1_I2NG_2016_M_22&format=json&start_date=2016-01-01&collapse=year&collapse_aggregation=avg"

#Con este enlace le estamos pidiendo los datos desde "start_date=" (si quisieramos un limite superior. podemos usar
#"end_date=")
#Con "collapse=year" le pedimos que la frecuencia temporal (tambien puede ser day, month, quarter, semester)
#Con "collapse_aggregation=avg" le decimos que hacer con la info que "collapse".
#En este caso le pedimos el promedio con avg, pero podriamos pedir la suma (sum),
#el minimo (min), maximo (max) o el ultimo valor (end_of_period)

response = requests.get(url_arg)
print(response)

datos = response.json()
datos
print("Datos extraídos desde 2016 con frecuencia anual (promedio):")
print(datos['data'])
```
